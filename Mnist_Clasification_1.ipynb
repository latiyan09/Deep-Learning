```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input, Flatten
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[0])
```

```python
y_train[0]
```

```python
x_train[0]
```

```python
x_train = x_train/225
x_test = x_test/225
```

```python
x_train[0]
```

```python
model = Sequential()
```

```python
model.add(Flatten(input_shape=(28,28)))
model.add(Dense(128, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(x_train, y_train, epochs=50, validation_split=0.2)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred = y_pred.argmax(axis=-1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```

### Model Prediction on a Sample Image

```python
# Select a sample image from the test set (e.g., the 5th image)
sample_index = 4
sample_image = x_test[sample_index]
sample_label = y_test[sample_index]

# Display the sample image
plt.imshow(sample_image)
plt.title(f"Actual Label: {sample_label}")
plt.show()
```

```python
# The model expects a batch of images, so we need to add an extra dimension
# Reshape the single image to (1, 28, 28)
input_image = sample_image.reshape(1, 28, 28)

# Make a prediction using the model
prediction_probabilities = model.predict(input_image)
predicted_label = prediction_probabilities.argmax(axis=-1)[0]

print(f"Predicted Label: {predicted_label}")
print(f"Actual Label: {sample_label}")
```
